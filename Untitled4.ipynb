```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
plt.figure(figsize=(8,4))
```

```python
data ={
    "Students": ["Aarav", "Priya", "Ananya", "Riya", "Arjun","Rahul", "Anjali","Divya", "Neha", "Ram"],
    "Marks":[85, 70, 89, 68, 50, 90, 93, 88, 60 , 94],
    "Attendance":[75, 87, 80, 60, 44, 90, 92, 83, 90, 97],
    "Study_hours":[4, 2, 4, 1, 2, 7, 8, 9, 4, 12]
    }
```

```python
df=pd.DataFrame(data)
df
```

```python
# BAR CHART
plt.bar(df["Students"], df["Marks"], color ="green")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.title("Marks of Students")
plt.xticks(rotation = 45)
plt.show()
```

```python
# HISTOGRAM
plt.hist(df["Marks"], bins=5, color="orange", edgecolor ="black")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.title("Marks Distribution")
plt.show()
```

```python
# SCATTER PLOT
plt.scatter(df["Study_hours"], df["Marks"], color="red", s=50, marker="o")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title("Study Hours vs Marks")
plt.show()
```

```python
# PIE CHART
plt.pie(df["Marks"], labels=df["Students"], autopct="%1.1f%%", startangle=90)
plt.axis("equal")
plt.title("Marks Distribution")
plt.show()
```

```python
# BOX PLOT
plt.boxplot(df["Marks"])
plt.ylabel("Marks")
plt.title("Marks Distribution")
plt.show()
```
